# 5. Fusione (merge) el dataframe de atletas con el de regiones usando la variable NOC, de modo que cada fila de atletas conserve su región.
# Verifique que el número de filas del dataframe de atletas no cambie después de la fusión y revise cuántas filas quedan sin región. ¿Por qué cree que ocurre esto? ¿Qué tipo de fusión (how) utilizó y por qué?

```python
df_merged = pd.merge(df_atletas, df_regiones, on='NOC', how='left')
# Esto indica que tome el dataframe izquierdo (df_atletas) y lo combine con el derecho (df_regiones)
# on='NOC' Es la columna que tienen ambos dataframes en comun. Se buscaran las filas donde el valor de 'NOC' coincida en ambas tablas para unirlas
# how='left' Conservarán absolutamente todos los registros del dataframe de la izquierda (df_atletas)
# Si el codigo de un atleta no existe en df_regiones no se eliminaro al atleta... Simplemente se pondra NaN
display(df_merged)
```

```python
# Comprobación de la cantidad de filas
print(len(df_merged))
print(len(df_atletas))
```

```python
# Cantidad de nulos en la columna de 'region'
print(df_merged['region'].isnull().sum())
```

#### ¿Por qué cree que ocurre esto?
Esto ocurre porque en el dataframe de atletas existen algunos equipos especiales o situaciones excepcionales que no están registrados formalmente en la tabla de diccionario de regiones (noc_regions.csv). Tras analizar los datos, encontramos que esos 370 registros nulos corresponden principalmente a ciertos valores de la columna NOC
#### ¿Qué tipo de fusión (how) utilizó y por qué?
Usamos `how = 'left'` porque es la que nos permite agregar las nuevas columnas del dataframe de regiones al final del otro dataframe, respetando los valores desconocidos del dataframe de atletas el cual es el que está a la izquierda en la línea de código `df_merged = pd.merge(df_atletas, df_regiones, on='NOC', how='left')`. Se mantiene ese orden en la fórmula para que no se alteren los valores del dataframe de atletas
